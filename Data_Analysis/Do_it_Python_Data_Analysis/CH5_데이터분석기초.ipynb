### 변수 조합해 파생 변수 만들기

````python
df['var_mean'] = (df['var1'] + df['var2']) / 2
df
````

### mpg 통합 연비 변수 만들기

````python
mpg['total'] = (mpg['cty'] + mpg['hwy']) / 2
mpg
````

````python
mpg['total'].mean()
````

### 조건문을 활용해 파생변수 만들기

````python
mpg['test'] = np.where(mpg['total'] >= 20, 'pass', 'fail')
mpg
````

````python
mpg['test'].value_counts()
````

````python
mpg['test'].value_counts().plot.bar()
````

````python
mpg['test'].value_counts().plot.bar(rot = 0) # x축 이름 회전
````

### 중첩 조건문 활용하기

````python
mpg['grade'] = np.where(mpg['total'] >= 30, 'A', 
                        np.where(mpg['total'] >= 20, 'B', 'C'))

mpg.head()
````

````python
mpg['grade'].value_counts()
````

````python
mpg['grade'].value_counts().sort_index().plot.bar(rot=0) # 알파벳순 정렬
````

````python
mpg['size'] = np.where((mpg['category'] == 'compact') |
                        (mpg['category'] == 'subcompact') |
                        (mpg['category'] == '2seater'),
                        'small', 'large')
mpg['size'].value_counts()
````

````python
mpg['size'] = np.where(mpg['category'].isin(['compact', 'subcompact', '2seater']), 'small', 'large') # 위의 코드와 동일한 결과
mpg['size'].value_counts()
````

### 문제 2. `poptotal`(전체인구) 변수를 `total`로, `popasian`(아시아 인구) 변수를 `asian`으로 수정하세요.

````python
midwest = midwest.rename(columns={'poptotal': 'total',
                'popasian': 'asian'})
midwest.head()
````

### 문제 3. `total`, `asian`변수를 이용해 '전체 인구 대비 아시아 인구 백분율' 파생변수를 추가하고 히스토그램을 만들어 분포를 살펴보세요.

````python
midwest['asian_ratio'] = midwest['asian'] / midwest['total'] *100
midwest['asian_ratio'].plot.hist()
````

### 문제 4. 아시아 인구 백분률 전체 평균을 구하고, 평균을 초과하면 `'large'`, 그 외에는 `'small'`을 부여한 파생변수를 만들어 보세요.

````python
asian_ratio_mean = midwest['asian_ratio'].mean()
print(asian_ratio_mean)

midwest['asian_size'] = np.where(midwest['asian_ratio']>asian_ratio_mean, 'large', 'small')
midwest['asian_size'].value_counts()
````

### 문제 5. `'large'`와 `'small'`에 대항하는 지역이 얼마나 많은지 빈도표와 빈도 막대 그래프를 만들어 확인해 보세요.

````python
midwest['asian_size'].value_counts().plot.bar(rot=0)
````
