## 6-6 집단별로 요약하기

### 집단별로 요약하기

```python
# math 평균 구하기
exam.agg(mean_math = ('math', 'mean'))
```

```python
# 집단별 요약 통계량 구하기
exam.groupby('nclass').agg(mean_math = ('math', 'mean'))
```

```python
# 변수를 인덱스로 바꾸지 않기
exam.groupby('nclass', as_index=False).agg(mean_math = ('math', 'mean'))
```

```python
# 여러 요약통계량 한 번에 구해가
exam.groupby('nclass', as_index=False)\
    .agg(mean_math = ('math', 'mean'),
         sum_math = ('math', 'sum'),
         median_math = ('math', 'median'),
         n = ('nclass' , 'count'))
```

`agg()`에 자주 사용되는 요약 통계량
|함수|통계량|
|---|---|
|`mean()`|평균|
|`std()`|표준편차|
|`sum()`|합계|
|`median()`|중앙값|
|`min()`|최솟값|
|`max()`|최댓값|

## 과제

미국 동북중부 437지역의 인구 통계 정보를 담고 있는 `midwest.csv`를 사용해 데이터 분석 문제를 해결해 보세요.

```python
midwest = pd.read_csv('data/midwest.csv')
midwest.head()
```

### 문제 1. `popadults`는 해당 지역의 성인 인구, `poptotal`은 전체 인구를 나타냅니다. `midwest`데이터에 '전체 인구 대비 미성년 인구 백분율' 변수를 추가하세요.

```python
midwest = midwest.assign(minor_pop_ratio = lambda x: (x['poptotal'] - x['popadults'])/x['poptotal'] * 100)
midwest.head()
```

### 문제 2. 미성년 인구 백분율이 가장 높은 상위 5개 `county`(지역)의 미성년 인구 백분율을 출력하세요.

```python
midwest.sort_values('minor_pop_ratio', ascending=False)[['county', 'minor_pop_ratio']].head(5)
```

### 문제 3. 분류표의 기준에 따라 미성년 비율 등급 변수를 추가하고, 각 등급에 몇 개의 지역이 있는지 알아보세요.
|분류|기준|
|---|---|
|large|40% 이상|
|middle|30~40% 이상|
|small|30% 미만|

```python
midwest = midwest.assign(minor_grade = lambda x: np.where(x['minor_pop_ratio'] >=  40, 'large', 
                                                np.where(x['minor_pop_ratio'] >= 30, 'middle', 'small')), inplace =True)

midwest['minor_grade'].value_counts()
```

### 문제 4. `popasian`은 해당 지역의 아시아인 인구를 나타냅니다. '전체 인구 대비 아시아인 인구 백분률' 변수를 추가하고 하위 10개 지역의 `state`(주), `county`(지역), 아시아인 인구 백분율을 출력하세요.

```python
midwest = midwest.assign(asian_ratio = lambda x: x['popasian']/x['poptotal']*100).sort_values('asian_ratio', ascending=True).reset_index()
midwest[['state', 'county', 'asian_ratio']].head(10)
```
