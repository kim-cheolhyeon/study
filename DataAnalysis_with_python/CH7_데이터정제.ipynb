```python
import pandas as pd
import numpy as np
```

## 7-1 이상치 제거하기

### 결측치 찾기

```python
# 결측치를 포함하는 데이터 생성
df = pd.DataFrame({'sex': ['M', 'F', np.nan, 'M', 'F'],
                   'score': [5, 4, 3, 4, np.nan]})
df
```

```python
# 결측치 확인하기
pd.isna(df)
```

```python
pd.isna(df).sum()
```

### 결측치 제거하기

```python
df.dropna(subset='score') # 하나의 속성에 대한 결측치 제거
```

```python
df.dropna(subset=['sex', 'score']) # 여러 개의 속성에 대한 결측치 제거
```

```python
df.dropna() # 모든 결측치가 있는 행 제거
```

### 결측치 대체하기

```python
exam = pd.read_csv('data/exam.csv')
exam.loc[[2, 7, 14], ['math']] = np.nan
exam
```

```python
mean_math = exam['math'].mean() # mean 매서드는 결측치가 있어도 작동
mean_math
```

```python
exam['math'] = exam['math'].fillna(mean_math) # 평균으로 결측치 대체하기
exam
```

```python
exam['math'].isna().sum()
```

## 7-2 이상치 정제하기

### 이상치 제거하기 - 존재할 수 없는 값

```python
df = pd.DataFrame({'sex': [1, 2, 1, 3, 2, 1], # 성별은 1과 2로 구분 3은 이상치
                   'score': [5, 4, 3, 4, 2, 6]}) # score는 1~5점을 가짐, 6은 이상치
df
```

```python
# 이상치 확인하기
df['sex'].value_counts()
```

```python
df['score'].value_counts()
```

```python
# 결측치로 변환하기
df['sex'] = np.where(df['sex']==3, np.nan, df['sex'])
df['score'] = np.where(df['score']>=5, np.nan, df['score'])
df
```
